```python
import sys
import os

sys.path.append(os.path.abspath(".."))

from Framework.data_loader import cargar_datos
from Framework.data_validation import validate_data
from Framework.data_cleaning import convertir_tipos
import pandas as pd
from Framework.data_cleaning import eliminar_duplicados
from Framework.data_cleaning import limpiar_nulos
```

```python
df = cargar_datos("../data/lumina_chocolate_supermercado.csv")

df.head()
```

```python
#Validación de datos antes de la conversión de tipos
validacion=validate_data(
    df,
    target="unidades_vendidas",
    fecha="fecha"
)

print(validacion)


#Selecciona las columnas que quieras convertir y el tipo de dato al que quieres convertirlas, nota la fecha se especificada en el parámetro fecha, no es necesario incluirla en el diccionario de conversión 
convertir={
    'precio_venta': 'float',
}

#Valida que los datos se convirtieron correctamente.
df = convertir_tipos(df, fecha="fecha",convertir=convertir)

validacion_tipos=validate_data(
    df,
    target="unidades_vendidas",
    fecha="fecha"
)

print(validacion_tipos)

#Eliminar Duplicados

df =eliminar_duplicados(df)





```

```python
df=limpiar_nulos(df, tratamiento ={'inversion_marketing' :'mediana',
                                     'distribucion_pct' : 'mediana',
                                     'precio_competidor' : 'interpolar'})

validate_data(df, target="unidades_vendidas", fecha="fecha")
```
